```python
import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

for pkg in ["punkt", "punkt_tab", "stopwords"]:
    nltk.download(pkg, quiet=True)

text = (
    "Technology has quietly rewritten the way I learn, build and think. "
    "Open source software is my favourite corner of it, because anyone can read the code, "
    "fix a bug and share the fix with the world. "
    "I'm especially fascinated by programming languages that catch mistakes before a program even runs, "
    "since they turn frustrating crashes into helpful compiler messages. "
    "It's amazing that a student with a laptop can contribute to tools used by millions of developers. "
    "Good technology doesn't just make tasks faster, it makes people more curious and more creative. "
    "That curiosity is what keeps me experimenting with new ideas every single day."
)
print("ORIGINAL TEXT:\n", text, "\n")

lower_text = text.lower()
clean_text = re.sub(r"[^\w\s]", "", lower_text)
print("CLEANED TEXT:\n", clean_text, "\n")

sentences = sent_tokenize(text)
print(f"SENTENCES ({len(sentences)}):")
for i, s in enumerate(sentences, 1):
    print(f"  {i}. {s}")

words = word_tokenize(clean_text)
print(f"\nWORD TOKENS ({len(words)}):\n", words, "\n")

split_tokens = lower_text.split()
nltk_tokens = word_tokenize(lower_text)

print("split()         ->", len(split_tokens), "tokens")
print("word_tokenize() ->", len(nltk_tokens), "tokens\n")

sample = "i'm especially fascinated by programming languages that catch mistakes"
print("Sample :", sample)
print("split()         :", sample.split())
print("word_tokenize() :", word_tokenize(sample), "\n")

print("Tokens only split() produced   :", sorted(set(split_tokens) - set(nltk_tokens)))
print("Tokens only word_tokenize() made:", sorted(set(nltk_tokens) - set(split_tokens)), "\n")

stop_words = set(stopwords.words("english"))
filtered = [w for w in words if w not in stop_words]
print(f"AFTER STOPWORD REMOVAL ({len(filtered)} of {len(words)} kept):\n", filtered, "\n")

fdist = FreqDist(filtered)
print("TOP 10 WORDS:")
for word, count in fdist.most_common(10):
    print(f"  {word:<15}{count}")

top = fdist.most_common(15)
labels, counts = zip(*top)
plt.figure(figsize=(10, 5))
plt.bar(labels, counts, color="steelblue")
plt.xticks(rotation=45, ha="right")
plt.title("Word Frequency Distribution (stopwords excluded)")
plt.xlabel("Word")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("word_frequency.png", dpi=150)
```

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

for pkg in ["stopwords", "wordnet", "omw-1.4"]:
    nltk.download(pkg, quiet=True)

text = (
    "Technology has quietly rewritten the way I learn, build and think. "
    "Open source software is my favourite corner of it, because anyone can read the code, "
    "fix a bug and share the fix with the world. "
    "I'm especially fascinated by programming languages that catch mistakes before a program even runs, "
    "since they turn frustrating crashes into helpful compiler messages. "
    "It's amazing that a student with a laptop can contribute to tools used by millions of developers. "
    "Good technology doesn't just make tasks faster, it makes people more curious and more creative. "
    "That curiosity is what keeps me experimenting with new ideas every single day."
)


words = re.findall(r"\b[a-z]+\b", text.lower())
print(f"1. ALPHABETIC WORDS ({len(words)}):\n", words, "\n")

stop_words = set(stopwords.words("english"))
filtered = [w for w in words if w not in stop_words]
print(f"2. AFTER STOPWORD REMOVAL ({len(filtered)}):\n", filtered, "\n")

stemmer = PorterStemmer()
stemmed = [stemmer.stem(w) for w in filtered]
print("3. STEMMED:\n", stemmed, "\n")


lemmatizer = WordNetLemmatizer()
lemma_noun = [lemmatizer.lemmatize(w) for w in filtered]
lemma_verb = [lemmatizer.lemmatize(w, pos="v") for w in filtered]
print("4a. LEMMATIZED (default, noun):\n", lemma_noun, "\n")
print("4b. LEMMATIZED (pos='v'):\n", lemma_verb, "\n")

print("5. COMPARISON (words where the outputs differ)")
print(f"{'original':<16}{'stem':<14}{'lemma (n)':<16}{'lemma (v)':<14}")
print("-" * 60)
for w, s, ln, lv in zip(filtered, stemmed, lemma_noun, lemma_verb):
    if len({w, s, ln, lv}) > 1:
        print(f"{w:<16}{s:<14}{ln:<16}{lv:<14}")

print("\nUnique vocabulary size:")
print("  original  :", len(set(filtered)))
print("  stemmed   :", len(set(stemmed)))
print("  lemma (n) :", len(set(lemma_noun)))
print("  lemma (v) :", len(set(lemma_verb)))
```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

texts = [
    "Apple launches new iPhone with faster chip and better battery life. "
    "Analysts say the new iPhone could boost sales this quarter.",
    "This laptop has a great screen and excellent battery life. "
    "The keyboard feels comfortable, but the fan is a little loud.",
    "India wins the cricket match by five wickets. "
    "The captain praised the bowlers after the match.",
]
labels = ["Text 1 (tech news)", "Text 2 (laptop review)", "Text 3 (cricket)"]

pd.set_option("display.width", 250)
pd.set_option("display.max_columns", None)

cv = CountVectorizer(stop_words="english")
bow = cv.fit_transform(texts)
bow_df = pd.DataFrame(bow.toarray(), index=labels, columns=cv.get_feature_names_out())
print("1. BAG OF WORDS (raw counts, stopwords removed)")
print(f"   Shape: {bow.shape}  (documents x vocabulary)\n")
print(bow_df.T, "\n")

tfidf = TfidfVectorizer(stop_words="english")
tfidf_matrix = tfidf.fit_transform(texts)
terms = tfidf.get_feature_names_out()
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), index=labels, columns=terms)
print("2. TF-IDF SCORES")
print(tfidf_df.T.round(3), "\n")

print("3. TOP 3 KEYWORDS PER TEXT (by TF-IDF)")
for label, row in tfidf_df.iterrows():
    top3 = row.sort_values(ascending=False).head(3)
    print(f"\n{label}")
    for term, score in top3.items():
        print(f"   {term:<12}{score:.3f}")
```

```python
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

for pkg in ["punkt", "punkt_tab", "stopwords"]:
    nltk.download(pkg, quiet=True)

text_ai = (
    "Artificial intelligence allows machines to learn from data and make decisions. "
    "Modern AI systems use neural networks trained on huge datasets to recognise images and understand language. "
    "Companies use AI for healthcare diagnosis, fraud detection and personalised recommendations. "
    "The quality of an AI model depends heavily on the data it learns from. "
    "Security and privacy of that data remain major concerns."
)
text_bc = (
    "Blockchain is a distributed ledger that records transactions across many computers. "
    "Each block is linked to the previous one using cryptography, which makes the records very hard to change. "
    "Cryptocurrencies like Bitcoin use blockchain to move money without a central bank. "
    "Smart contracts let agreements run automatically on the network. "
    "Security and transparency are the main strengths of the technology."
)


stop_words = set(stopwords.words("english"))

def preprocess(text):
    text = re.sub(r"[^\w\s]", "", text.lower())
    tokens = word_tokenize(text)
    return [t for t in tokens if t not in stop_words]

tokens_ai = preprocess(text_ai)
tokens_bc = preprocess(text_bc)
print("AI tokens        :", tokens_ai)
print("Blockchain tokens:", tokens_bc, "\n")


def jaccard(a, b):
    a, b = set(a), set(b)
    return len(a & b) / len(a | b)

set_ai, set_bc = set(tokens_ai), set(tokens_bc)
print("2a. JACCARD SIMILARITY")
print("   |A| =", len(set_ai), " |B| =", len(set_bc))
print("   Shared words :", sorted(set_ai & set_bc))
print("   |A ∩ B| =", len(set_ai & set_bc), " |A ∪ B| =", len(set_ai | set_bc))
print(f"   Jaccard = {jaccard(tokens_ai, tokens_bc):.4f}\n")

clean_ai = " ".join(tokens_ai)
clean_bc = " ".join(tokens_bc)
tfidf = TfidfVectorizer()
matrix = tfidf.fit_transform([clean_ai, clean_bc])
cos = cosine_similarity(matrix[0], matrix[1])[0][0]
print("2b. COSINE SIMILARITY (TF-IDF)")
print(f"   Cosine = {cos:.4f}\n")


raw_ai = re.sub(r"[^\w\s]", "", text_ai.lower()).split()
raw_bc = re.sub(r"[^\w\s]", "", text_bc.lower()).split()
raw_cos_nostop = cosine_similarity(
    TfidfVectorizer(stop_words=None).fit_transform([" ".join(raw_ai), " ".join(raw_bc)])
)[0][1]
print("2c. EFFECT OF STOPWORD REMOVAL")
print(f"   Jaccard  without stopword removal: {jaccard(raw_ai, raw_bc):.4f}   with: {jaccard(tokens_ai, tokens_bc):.4f}")
print(f"   Cosine   without stopword removal: {raw_cos_nostop:.4f}   with: {cos:.4f}\n")
```

```python
import pandas as pd
from textblob import TextBlob
from wordcloud import WordCloud, STOPWORDS
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

reviews = [
    "These earbuds are fantastic! The sound quality is excellent and the battery lasts all day.",
    "Absolutely love them. Comfortable fit, great bass and the noise cancellation works beautifully.",
    "Amazing value for the price. Setup was easy and the sound is crisp and clear.",
    "Terrible experience. The left earbud stopped working after two weeks and support was useless.",
    "The battery drains quickly and the connection keeps dropping. Very disappointing and frustrating.",
    "Poor build quality, the case feels cheap and the sound is awful.",
    "The earbuds arrived on Tuesday in a small white box. They come with three sizes of ear tips.",
    "Good battery life and a comfortable design, but the touch controls are a bit annoying.",
    "Wonderful sound and a stylish design. Best earbuds I have owned, highly recommended!",
    "The package includes a charging case and a USB-C cable. Bluetooth version is 5.3.",
]

rows = []
for r in reviews:
    s = TextBlob(r).sentiment
    rows.append({"review": r, "polarity": round(s.polarity, 3), "subjectivity": round(s.subjectivity, 3)})
df = pd.DataFrame(rows)

def classify(p, threshold=0.1):
    if p > threshold:
        return "Positive"
    if p < -threshold:
        return "Negative"
    return "Neutral"

df["sentiment"] = df["polarity"].apply(classify)

show = df.copy()
show["review"] = show["review"].str.slice(0, 45) + "..."
print(show[["review", "polarity", "subjectivity", "sentiment"]].to_string(), "\n")
print("Counts:\n", df["sentiment"].value_counts(), "\n")

positive_text = " ".join(df.loc[df["sentiment"] == "Positive", "review"])
extra_stop = {"earbuds", "one", "will", "also"}
wc = WordCloud(
    width=900, height=450, background_color="white",
    stopwords=STOPWORDS | extra_stop, colormap="viridis", random_state=42,
).generate(positive_text)

plt.figure(figsize=(10, 5))
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud - Positive Reviews")
plt.tight_layout()
plt.savefig("positive_wordcloud.png", dpi=150)
plt.show()
```
