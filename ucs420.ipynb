```python
roll_number = "1024170184"
digits=[int(d) for d in roll_number[:8]]
print("digits are",digits)
```

```python
A={d*7 for d in digits}
B={d*9 for d in digits}

print("Set A",A)
print("Set B",B)
```

```python
union_set=A.union(B)
print("union set",union_set)
```

```python
# vii. Intersection
intersection_AB = A.intersection(B)
print("Intersection of A and B:", intersection_AB)

# viii. Difference
diff_A_minus_B = A.difference(B)
diff_B_minus_A = B.difference(A)
print("A - B:", diff_A_minus_B)
print("B - A:", diff_B_minus_A)
print("Note: difference() gives elements in one set but not the other (one-directional, "
      "so A-B and B-A differ), while symmetric_difference() gives all elements that are "
      "in either set but NOT in both — i.e., the union of A-B and B-A combined.")

# ix. Symmetric difference
sym_diff = A.symmetric_difference(B)
print("\nSymmetric difference of A and B:", sym_diff)

# x. Subset / Superset checks
print("\nIs A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

# xi. Remove user input value from A using discard()
X = int(input("\nEnter a value X to remove from set A: "))
A.discard(X)
print("Set A after discard():", A)
print("Note: discard() is safer than remove() because discard() does nothing if X is "
      "not present, whereas remove() raises a KeyError, which can crash the program.")
```
